```python
# ввод числа N
N = int(input())

# здесь задается функция fib_rec (переменную N не менять!)
def fib_rec(N, f=[1, 1]):
    if N > 1:
        f.append(f[-2] + f[-1])
        fib_rec(N-1, f)
    return f
print(fib_rec(8, f=[1, 1]))

```
