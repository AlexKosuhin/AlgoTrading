## 📌 Основные служебные атрибуты и функции классов Python

### 🧠 `__dict__` — все атрибуты в словаре  
• У объекта: только его собственные атрибуты  
• У класса: его атрибуты + служебные данные  

```python
class Cat:
    species = "кошка"

c = Cat()
c.name = "Мурка"

print(c.__dict__)      # {'name': 'Мурка'}
print(Cat.__dict__)    # содержит 'species', '__doc__', и др.
```

---

### 📖 `__doc__` — документация класса  
• Хранит первую строку-комментарий после названия класса  

```python
class Dog:
    """Это класс собаки"""
    pass

print(Dog.__doc__)  # Это класс собаки
```

---

### 🔍 `getattr()` — безопасное чтение  
• Если атрибута нет и **нет третьего аргумента** — будет ошибка `AttributeError`  
• Третий аргумент — значение «на всякий случай»  

```python
class Car:
    color = "синий"

c = Car()

print(getattr(c, 'color'))         # синий  
print(getattr(c, 'speed', 0))      # 0 (speed не существует)  
# print(getattr(c, 'speed'))       # ОШИБКА: AttributeError
```

---

### ✏️ `setattr()` — установка атрибута по имени  
• Делает то же самое, что `объект.имя = значение`, но имя — строка  

```python
class Box:
    pass

b = Box()
setattr(b, 'size', 10)   # то же, что b.size = 10
print(b.size)            # 10
```

---

### ❓ `hasattr()` — проверка, есть ли атрибут  
• Возвращает `True` или `False` — ничего не ломается  

```python
class Bird:
    wings = 2

b = Bird()

print(hasattr(b, 'wings'))   # True  
print(hasattr(b, 'legs'))    # False
```

---

### 🗑️ `delattr()` — удаление атрибута  
• Удаляет атрибут по имени  
• Если атрибута нет — будет ошибка  

```python
class Toy:
    price = 100

t = Toy()
t.color = "красный"

delattr(t, 'color')        # удалили color
# print(t.color)           # ОШИБКА: такого атрибута больше нет

delattr(Toy, 'price')      # удалили атрибут класса
```

---

### 🧪 `type()` и `isinstance()` — проверка типа  
• `type()` показывает, от какого класса объект  
• `isinstance()` — правильный способ проверить тип  

```python
class Apple:
    pass

a = Apple()

print(type(a) == Apple)       # True  
print(isinstance(a, Apple))   # True ← лучше использовать это
```

---

### 🔁 Приоритет поиска атрибутов  
• Сначала смотрит у объекта, потом — у класса  

```python
class Pen:
    color = "чёрный"

p = Pen()
print(p.color)        # чёрный (из класса)

p.color = "синий"     # теперь у объекта свой color
print(p.color)        # синий

del p.color           # удалили у объекта
print(p.color)        # снова чёрный (из класса)
```

---

### 💎 Главное правило  
> 📊 **"Когда читаешь — ищет везде. Когда пишешь — записывает только в объект."**  
> Поэтому у каждого объекта могут быть свои значения, даже если в классе есть такой же атрибут.

### 📝 Краткий конспект: **Методы классов в Python и параметр `self`**

---

#### 🔹 **Методы vs Свойства**
- **Свойства (данные)** → имена **существительные**:  
  ```python
  color, size, x, y
  ```
- **Методы (функции)** → имена **глаголы**:  
  ```python
  set_coords, get_param, start, stop
  ```

---

#### 🔹 **Базовое объявление метода**
```python
class Point:
    color = 'red'
    circle = 2

    def set_coords(self):               # ← self обязателен для вызова из экземпляра
        print("вызов метода set_coords")
```

- `self` — **автоматическая ссылка на экземпляр**, из которого вызван метод.
- Если метод вызывается **напрямую через класс**, `self` нужно передать вручную:
  ```python
  pt = Point()
  pt.set_coords()                     # ✅ OK — self = pt (автоматически)
  Point.set_coords()                  # ❌ Ошибка — self не передан
  Point.set_coords(pt)                # ✅ OK — self = pt (вручную)
  ```

---

#### 🔹 **Зачем нужен `self`?**
Позволяет методу работать **с конкретным экземпляром**:

```python
def set_coords(self, x, y):
    self.x = x      # создаёт/меняет атрибут x в ЭКЗЕМПЛЯРЕ
    self.y = y      # а не в классе!
```

Пример:
```python
pt1 = Point()
pt2 = Point()

pt1.set_coords(1, 2)     # → pt1.x = 1, pt1.y = 2
pt2.set_coords(10, 20)   # → pt2.x = 10, pt2.y = 20

print(pt1.__dict__)      # {'x': 1, 'y': 2}
print(pt2.__dict__)      # {'x': 10, 'y': 20}
```
→ Атрибуты создаются **в пространстве имён объекта**, а не класса.

---

#### 🔹 **Пример: методы для установки и получения данных**
```python
class Point:
    color = 'red'
    circle = 2

    def set_coords(self, x, y):
        self.x = x
        self.y = y

    def get_coords(self):
        return (self.x, self.y)   # обращение к атрибутам экземпляра
```

Использование:
```python
pt = Point()
pt.set_coords(5, 7)
print(pt.get_coords())   # (5, 7)
```

---

#### 🔹 **Методы — тоже атрибуты!**
Можно получать методы динамически:
```python
res = getattr(pt, 'get_coords')   # ← ссылка на метод
print(res())                      # (5, 7)
```
→ Имя метода — это просто **атрибут-функция**.

---

#### ✅ **Итог**
| Факт | Объяснение |
|------|-------------|
| `self` — 1-й параметр | Автоматически ссылается на экземпляр при вызове `obj.method()` |
| Методы общие для всех объектов | Хранятся в классе, но работают с конкретным `self` |
| Атрибуты создаются в `self` | `self.x = ...` → запись в `__dict__` экземпляра |
| Вызов через класс | Требует явной передачи `self`: `Class.method(obj)` |



> 🎯 `self` — ключ к индивидуальности объектов: позволяет каждому экземпляру хранить **свои** данные, используя **общие** методы.


# Конспект: Инициализация и финализация объектов в Python

## 🧙‍♂️ Магические методы
— Специальные методы, название которых начинается и заканчивается двойным подчёркиванием (`__method__`).  
— Автоматически вызываются интерпретатором при определённых событиях.

---

## 🚀 `__init__(self, ...)`
**Назначение:** инициализатор — выполняет начальную настройку объекта сразу после его создания.

### Особенности:
- Вызывается **автоматически** после `__new__`, если объект успешно создан.
- Параметр `self` ссылается на созданный экземпляр.
- Может принимать дополнительные аргументы (как обычная функция):
  ```python
  def __init__(self, x=0, y=0):
      self.x = x
      self.y = y
  ```
- Позволяет задавать начальные значения атрибутов при создании:
  ```python
  pt = Point()        # x=0, y=0
  pt = Point(10)      # x=10, y=0
  pt = Point(1, 2)    # x=1, y=2
  ```

> 💡 Без `__init__` объект создаётся «пустым» — атрибуты нужно задавать вручную.

---

## 🗑️ `__del__(self)`
**Назначение:** финализатор — вызывается **перед удалением** объекта.

### Особенности:
- Вызывается автоматически **сборщиком мусора**, когда на объект больше нет внешних ссылок.
- Не гарантируется точное время вызова (зависит от реализации GC).
- Пример:
  ```python
  def __del__(self):
      print(f"Удаление экземпляра: {self}")
  ```

> ⚠️ Не следует полагаться на `__del__` для критически важной очистки ресурсов (файлы, соединения и т.п.) — лучше использовать `with` и контекстные менеджеры.

---

## 📌 Итог
| Метод         | Когда вызывается           | Для чего используется         |
|---------------|----------------------------|-------------------------------|
| `__init__`    | После создания объекта     | Инициализация атрибутов       |
| `__del__`     | Перед удалением объекта    | Очистка (не для критичных ресурсов) |

✅ Понимание этих методов — ключ к управлению жизненным циклом объектов в Python.
