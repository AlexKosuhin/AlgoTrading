```python
prices = np.random.normal(100, 5, 100)

def rsi(prices, n=14):
    deltas = np.diff(prices)
    gains = np.where(deltas > 0, deltas, 0)
    loss = np.where(deltas < 0, -deltas, 0)
    
    # Расчитываем первые значения
    avg_gain = np.mean(gains[:n])
    avg_loss = np.mean(loss[:n])
    
    rsi = np.zeros_like(prices)
    rsi[:n] = np.nan
    
    for i in range(n, len(prices)):
        avg_gain = (avg_gain * (n - 1) + gains[i-1]) / n
        avg_loss = (avg_loss * (n - 1) + loss[i-1]) / n
        rs = avg_gain / avg_loss
        rsi[i] = 100 - 100 / (1 + rs)
    return rsi

rsi_result = rsi(prices, 14)
sns.lineplot(rsi_result);
```

# MACD

```python
def EMA(prices, n):
    ema_values = np.zeros_like(prices)
    a = 2 / (n + 1)
    ema_values = np.full_like(prices, fill_value=0, dtype=float)
    ema_values[n-1] = np.mean(prices[:n])
    for i in range(n, len(prices)):
        ema_values[i] = a * prices[i] + (1 - a) * ema_values[i-1]
    return ema_values
```

```python
# MACD = EMA(12) - EMA(26)
# Signal = EMA(MACD, 9)
# Histogram = MACD - Signal
np.random.seed(32)
prices = np.random.normal(100, 2000, 400)
prices_ = np.cumsum(prices)
prices_s = pd.Series(prices_)

EMA_12 = EMA(prices_s, 12)
EMA_26 = EMA(prices_s, 26)
MACD = EMA_12 - EMA_26
Signal = EMA(MACD, 9)
Histogram = MACD - Signal
fig, axes = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

# Графики цен и EMA
axes[0].plot(prices_s, label='Цена')
axes[0].plot(EMA_12, label='EMA 12')
axes[0].plot(EMA_26, label='EMA 26')
axes[0].set_title("Цена и EMA")
axes[0].legend()
axes[0].grid(True)

# Графики MACD и Signal
axes[1].plot(MACD, label='MACD')
axes[1].plot(Signal, label='Signal')

# Гистограмма
colors = ['g' if val >= 0 else 'r' for val in Histogram]
axes[1].bar(range(len(Histogram)), Histogram, color=colors, alpha=0.5)

axes[1].set_title("MACD, Signal и Гистограмма")
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

Задача 8: Bollinger Bands
Условие: Реализуйте полосы Боллинджера (SMA 20, ±2 стандартных отклонения).

```python
prices = np.random.normal(0, 100, 500)
prices_cum = np.cumsum(prices)
def Bollinger_Bands(prices, n=20):
    prices = pd.Series(prices)
    SMA = prices.rolling(n).mean()
    std = prices.rolling(n).std()
    SMA_upper = SMA + 2 * std
    SMA_down = SMA -2 * std 
    return prices, SMA, SMA_upper, SMA_down

prices, SMA, SMA_upper, SMA_down = Bollinger_Bands(prices_cum)
# Графики линии
plt.figure(figsize=(10, 8))
sns.lineplot(data=prices, label='Prices')
sns.lineplot(data=SMA, label='SMA')
sns.lineplot(data=SMA_upper, label='SMA Upper')
sns.lineplot(data=SMA_down, label='SMA Down')
plt.title('График цен и SMA')
plt.xlabel('Временной индекс')
plt.ylabel('Значение')
plt.legend(title='Легенда')
plt.grid(True)
plt.tight_layout()
plt.show()
```
